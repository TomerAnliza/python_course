#🟢 Beginner

Exercise 1

Create and print a dictionary

Create a dictionary with 3 key-value pairs representing fruits and their colors and print it.



```
The color of apple is red

The color of banana is yellow

The color of orange is orange
```





Exercise 2

Access elements

Print the color of one specific fruit from the dictionary.



Exercise 3

Add elements

Add a new fruit with its color to the dictionary and print the updated dictionary.



Exercise 4

Remove elements

Remove a fruit from the dictionary by its key and print the updated dictionary.



#🟡 Intermediate

Exercise 5

Keys and values

Print all the keys and then all the values from the dictionary.



Exercise 6

Loop through dictionary

Use a for loop to print each fruit and its color in the format: "Fruit: Color".



Exercise 7

Update values

Change the color of one fruit in the dictionary and print the updated dictionary.
