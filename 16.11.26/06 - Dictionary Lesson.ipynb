## Creating a new `dict`

````python
my_dict = {
    'name' : 'banana'
}
my_dict
````

## Adding a value

````python
my_dict['color'] = 'green'
my_dict['weight'] = 120
print(my_dict)
````

## Updating a value

````python
my_dict['color'] = 'yellow'
print(my_dict)
````

*italicized text*## Accessing a value using the key

````python
my_dict['name']
````

## Accessing all values using `.values()` method

````python
'banana' in my_dict.values()
````
