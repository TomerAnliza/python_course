Abstraction
1. Add The Square shape
2. add the perimeter feature

```python
from abc import ABC, abstractmethod


class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass


class Square(Shape):
    def __init__(self, a):
        self.aa = a

    def area(self):
        return self.aa**2

    def perimeter(self):
        return 4 * self.aa


class Circle(Shape):
    def __init__(self, r):
        self.r = r

    def area(self):
        return 3.14 * self.r * self.r

    def perimeter(self):
        return 2 * 3.14 * self.r


class Rectangle(Shape):
    def __init__(self, l, w):
        self.length = l
        self.width = w

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)


shape = Circle(5)
print(shape.area())  # 78.5
get_rect = Rectangle(4, 6)
print(get_rect.perimeter())  # 24
my_square = Square(10)
print(my_square.area())
print(my_square.perimeter())
```
