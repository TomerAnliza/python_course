## Demo args

```python
def my_function(*kids):
  print("The youngest child is " + kids[2])
  print(type(kids))

my_function("Emil", "Tobias", "Linus", "John", "Jane", "Joe", "Jill", "Jack", "Jerry", "Jasmine")
```

```python
def my_function(*numbers):
    total = 0
    for num in numbers:
        total += num
    return total


print(my_function(1, 2, 3))
print(my_function(10, 20, 30, 40))
print(my_function(5))

# print(sum((1, 2, 3)))
```

```python
def my_function(*numbers):
  if len(numbers) == 0:
    return None
  max_num = numbers[0]
  for num in numbers:
    if num > max_num:
      max_num = num
  return max_num

print(my_function(3, 7, 2, 9, 1))

# print(max((1, 2, 3, 4, 5))) # returns 5
```

Kwargs = keyword arguments. **kwargs

```python
def my_function(**kid):
  print("His last name is " + kid.get("lname", "unknown"))

  print(type(kid))
  print(kid)

my_function(fname = "Tobias", lname = "Refsnes")
```

Unpacking a list/tuple

```python
lst = [1, 2, 3]
lst2 = [4, 5, 6]

# shallow copy of lst1 and lst2
combo = [*lst, *lst2]

print(combo)
```

unpacking dictionaries:

```python
def my_function(fname, lname):
  print("Hello", fname, lname)

# my_function(fname="dave", lname="woot")

person = {"fname": "Emil", "lname": "Refsnes"}


my_function(**person) # Same as: my_function(fname="Emil", lname="Refsnes")
```

Lambda functions:
Syntax: 
        lambda arguments : expression


```python
x = lambda a: a + 10
print(x(1))
```

```python
x = lambda a, b : a * b
print(x(5, 6))
```

built-in functions:

map:

```python
numbers = [1, 2, 3, 4, 5]

doubled = list(map(lambda x: x * 2, numbers))

# list comprehension:
doubled = [x * 2 for x in numbers]

print(doubled)
```

filter:

```python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
odd_numbers = list(filter(lambda x: x % 2 != 0, numbers))
print(odd_numbers)

# list comprehension with a condition:
# odd = [x for x in numbers if x % 2 != 0]
```

sorted:

```python
lst = [1, 10, 19, 3, 1, 5]
lst_sorted = sorted(lst)
print(lst)

students = [("Emil", 25), ("Tobias", 22), ("Linus", 28)]
sorted_students = sorted(students, key=lambda x: x[1])
print(sorted_students)
```
