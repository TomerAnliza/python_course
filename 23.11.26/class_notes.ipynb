```python
def upper_cased(func):
    def myinner():
        return func().upper()

    return myinner

@upper_cased
def myfunction():
    return "Hello Sally"


print(myfunction())
```

```python
from functools import wraps


def changecase(func):
  @wraps(func)
  def myinner():
    return func().upper()
  return myinner

@changecase
def myfunction():
  return "Have a great day!"

print(myfunction.__name__)
```
